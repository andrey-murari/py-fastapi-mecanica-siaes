```python
from datetime import date

from pydantic import BaseModel
from pydantic import ValidationError
```

```python
class ClienteSchema(BaseModel):
    id: int
    nome: str
    sobrenome: str
    email: str
    data_nascimento: date
```

```python

input = {
    'id': 1,
    'nome': 'João',
    'sobrenome': 'Silva',
    'email': 'joao.silva@email.com',
    'data_nascimento': '2026-06-01'
}
```

```python
try:
    cliente = ClienteSchema(**input)
    print(cliente.nome)
    print(cliente.model_dump_json())
except ValidationError as e:
    print(e.errors())
```
